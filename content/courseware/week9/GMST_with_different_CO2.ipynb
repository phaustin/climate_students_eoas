```python
# Snowball Cell 1
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import climlab
from climlab import constants as const
from climlab import legendre
```

```python
# create model
model3 = climlab.EBM_annual( num_points = 180, a0=0.3, a2=0.078, ai=0.62) # Snowball Cell 2
LW_CO2_3 = climlab.radiation.AplusBT_CO2(CO2=400, state=model3.state, **model3.param) #? https://climlab.readthedocs.io/en/latest/api/climlab.radiation.AplusBT.html
# add CO2 process
model3.add_subprocess('LW', LW_CO2_3) #https://climlab.readthedocs.io/en/latest/api/climlab.radiation.AplusBT.html

# run to equilibrium
model3.integrate_years(20)

# set CO2 concentration
model3.subprocess["LW"].CO2 = 10

temps = [climlab.global_mean(model3.Ts)]
times = [0]
# get GMST after 0.2 year time steps
for i in range(50):
    model3.integrate_years(0.2)
    temps.append(climlab.global_mean(model3.Ts))
    times.append(i+1)
# put the times in units of years
times = np.array(times)/5
# create plot
plt.plot(times, temps)
plt.xlabel("Year")
plt.ylabel("Global mean surface temperature (deg C)")
plt.title(f'Global temperature over time when CO2 concentration = {model3.subprocess["LW"].CO2} ppm')
```
